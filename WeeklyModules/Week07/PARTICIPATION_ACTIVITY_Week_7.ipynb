```python
!git clone https://github.com/239-blip/mssp607
%cd /content/mssp607
import pandas as pd
df = pd.read_csv('WeeklyModules/Week07/InternetSales.csv',encoding='cp1252')
```

```python
import matplotlib.pyplot as plt
df['Year']=pd.to_datetime(df['OrderDate']).dt.year

#calculate total sales by years
yearly_sales=df.groupby('Year')['SalesAmount'].sum()

#create figure and subplot
fig=plt.figure()
ax = fig.add_subplot(1,1,1)

#create bar chart
yearly_sales.plot.bar(ax=ax, color='k', alpha=0.8)

#add labels
ax.set_title('Yearly Internet Sales')
ax.set_xlabel('Year')
ax.set_ylabel('Total Sales')

#save as PDF
fig.savefig('Yearly_Sales.pdf')

plt.show()
```

```python
#read data from GitHub
!git clone https://github.com/239-blip/mssp607
%cd mssp607
import pandas as pd
aapl=pd.read_csv("data/aapl.csv")
amzn = pd.read_csv("data/amzn.csv")
fb = pd.read_csv("data/fb.csv")
goog = pd.read_csv("data/goog.csv")
nflx = pd.read_csv("data/nflx.csv")
```

```python
#c. add a ticker column
aapl["ticker"] = "AAPL"
amzn["ticker"] = "AMZN"
fb["ticker"] = "FB"
goog["ticker"] = "GOOG"
nflx["ticker"] = "NFLX"
```

```python

#d. combine
faanf = pd.concat([fb, aapl, amzn, nflx, goog],
                  ignore_index=True
)
```

```python
#e. save the result
faanf.to_csv("faanf.csv", index=False)
```

```python
#f
faanf = pd.read_csv("faanf.csv")
#convert to datetime
faanf["date"] = pd.to_datetime(faanf["date"])
# convert to integer
faanf["volume"] = faanf["volume"].astype(int)

faanf = faanf.sort_values(["date", "ticker"])
```

```python
# g
top7_vol = faanf.nlargest(7, "volume")

print(top7_vol)
```

```python
# h. melt
melted = faanf.melt(
    id_vars=["date", "ticker"],
    var_name="variable",
    value_name="value"
)

print(melted)
```
